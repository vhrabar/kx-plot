## Groups with `by=`

`kde`, `ridge`, `ecdf`, `box`, `violin`, `strip` and `qq` split `x` into one series per group.

```python
long = dists.melt(var_name='dist', value_name='value')
for k in ('kde', 'ridge', 'ecdf', 'box', 'violin', 'strip', 'qq'):
    kx.plot(long.value, spec=k, by=long.dist, title=f'{k}, by=dist')
```

## Palettes and colourmaps

`kx.use(theme, palette=..., cmap=...)`; add `_r` to any colourmap to reverse it.

```python
for pal, cmap in (('tollight', 'batlow'), ('set1', 'thermal'), ('paired', 'vik_r'), ('tab20', 'romaO')):
    kx.use('dark', palette=pal, cmap=cmap)
    kx.plot(cloud.x, cloud.y, cloud.z, 'scatter', title=f'{pal} + {cmap}')
    kx.plot(dists.normal, dists.skewed, dists.bimodal, 'kde-leg', title=pal)
kx.use('dark')
```

## Machine learning

Truth first, then the scores: `kx.plot(y_true, score, kind)`. A second model goes in `z`, a 2-D score draws one curve per class.

```python
binary, multi, reg, history, imp, curves = (kx.load(n) for n in ('binary', 'multiclass', 'regression', 'history',
                                                                    'importance', 'curves'))
```

```python
for k in ('roc', 'pr', 'calib', 'gain', 'lift'):
    kx.plot(binary, 'y', 'gbm', 'logreg', k, title=k)
```

```python
kx.plot(binary, 'y', 'gbm', 'proba-norm', title='proba-norm');
kx.plot(binary, 'y', 'gbm', 'thresh-best', title='thresh-best');
```

```python
kx.plot(multi.y, multi[['cat', 'dog', 'bird']], 'roc', title='roc, one curve per class');
kx.plot(multi, 'y', 'pred', 'cm-norm-ann', title='cm-norm-ann');
```

```python
kx.plot(reg, 'actual', 'predicted', 'pva', title='pva');
kx.plot(reg, 'actual', 'predicted', 'x', 'resid', title='resid, coloured by x');
```

```python
kx.plot(imp, 'feature', 'importance', 'std', 'imp', top=10, title='imp, top=10');
kx.plot(history, 'loss-best', title='loss-best');
kx.plot(curves['size'], curves.filter(like='train_'), curves.filter(like='val_'), 'lcurve', title='lcurve');
```

```python
ks = np.arange(1, 11)
kx.plot(ks, 1000 / ks ** 1.2 + 40, pd.Series(0.6 - 0.04 * np.abs(ks - 3), name='silhouette'), 'elbow-best', title='elbow-best');
```

```python
kx.plot(reg, 'actual', 'predicted', 'scatter-diag-fit', title='scatter-diag-fit');
```

```python
kx.evaluate(binary.y, binary.gbm, binary.logreg);
```

## Profiles, panels and statistics

`radar` and `parallel` compare groups across columns; `pair` and `cluster` draw several panels; `facet` splits any plot by a column; `sig` marks the groups that differ.

```python
profile = passengers[['age', 'fare', 'pclass', 'survived', 'sex']]
kx.plot(profile, 'radar', by='sex', title='radar, by=sex');
kx.plot(profile, 'parallel-mean', by='sex', title='parallel-mean');
```

```python
embedding = kx.load('embedding')
kx.plot(embedding, 'pair', by='label', cols=['x', 'y', 'size', 'f0'], title='pair');
```

```python
kx.plot(embedding.select_dtypes('number'), 'cluster-corr', title='cluster-corr');
```

```python
kx.plot(passengers, 'pclass', 'port', 'crosstab-facet-ann', facet='sex', title='crosstab-facet');
```

```python
kx.plot(passengers, 'fare', spec='box-sig', by='pclass', title='box-sig: fare by class');
```

```python
kx.cramers(passengers, cluster=True, title='kx.cramers');
```

## Images

`kx.load('shapes')` is 210 small images in 7 classes, with masks, a model's predictions and saliency maps.

```python
shapes = kx.load('shapes')
kx.plot(shapes['images'], shapes['labels'], 'imgrid', title='imgrid');
```

```python
kx.plot(shapes['images'], shapes['labels'], 'meanimg', title='meanimg');
kx.plot(shapes['images'], shapes['labels'], 'meanimg-std', title='meanimg-std');
```

```python
kx.plot(shapes['images'], shapes['labels'], 'colorhist', title='colorhist');
kx.plot(shapes['images'], 'imhist', idx=[0, 1], title='imhist');
```

```python
pred = np.roll(shapes['masks'], 2, axis=2)   # a prediction 2 pixels off
kx.plot(shapes['images'], shapes['masks'], pred, 'segmask', n=3, title='segmask');
```

```python
kx.plot(shapes['images'], shapes['saliency'], 'saliency', n=3, title='saliency');
kx.plot(shapes['images'], shapes['labels'], shapes['proba'], 'miscls', n=9, title='miscls');
```
